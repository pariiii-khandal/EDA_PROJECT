```python
import numpy as np
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import LabelEncoder

# 1. Load your dataset
df = pd.read_csv("fraudTrain.csv")

# 2. Select numerical columns & encode categorical columns for the model
# We drop unique identifiers and text features that cause noise
selected_features = ["category", "amt", "gender", "city_pop", "lat", "long"]
X = df[selected_features].copy()

# Convert categorical features (category, gender) into numerical codes
le = LabelEncoder()
X["category"] = le.fit_transform(X["category"])
X["gender"] = le.fit_transform(X["gender"])

# 3. Handle the true target label
target = df["is_fraud"]

# 4. Initialize the Isolation Forest model
# Fraud is usually rare (~0.2% to 0.5% in this specific dataset format)
model = IsolationForest(contamination=0.005, random_state=42)

# 5. Fit the model and predict anomalies (-1 for fraud, 1 for legitimate)
df["anomaly_score"] = model.fit_predict(X)

# 6. Map predictions to match your 'is_fraud' label (1 = Fraud, 0 = Legitimate)
df["predicted_fraud"] = np.where(df["anomaly_score"] == -1, 1, 0)

# 7. Evaluate results
matches = np.sum(df["predicted_fraud"] == target)
print(f"Total correct matching predictions: {matches}")
```
